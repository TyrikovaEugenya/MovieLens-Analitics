## Мультижанры
Бывает, смотришь фильм и не можешь понять — это комедия, драма или фантастика? Некоторые картины ловко смешивают кучу жанров. Давай посмотрим на десяток таких

```python
%timeit movies.most_genres(10)
movies.most_genres(10)
```

Кто-то снимает редко, но метко, а кто-то выдаёт хит за хитом без остановки. top_directors() вернет самых продуктивных. Возможно, твой любимый режиссёр тоже в списке

```python
%timeit links.top_directors(10)
links.top_directors(10)
```

Ты когда-нибудь задумывался, во сколько обходится создание крутого голливудского кино(и не только голливудского)? Этот метод соберет 10 фильмов, на которые потратили кучу денег. Бюджеты тут сумашедшие - сам посмотри. 


```python
%timeit links.most_expensive(10)
links.most_expensive(10)
```

Для этих фильмов пачка чипсов или попкорна потребуется побольше. Если вдруг есть свободных 3-4 часа, вызывай longest() и начинай смотреть по порядку

```python
%timeit links.longest(10)
links.longest(10)
```

Хочешь узнать, какие фильмы оказались самыми прибыльными? Метод ниже покажет 10 фильмов, где разница между сборами и бюджетом максимальна. 

```python
%timeit links.most_profitable(10)
links.most_profitable(10)
```

А теперь прикинем, сколько стоит каждая минута фильма!

## Тeги

Теги помогут тебе понять, о чём фильм, даже без прочтения описания. 

Может и не самая полезная выборка, но лучше посмотреть на самые длинные теги, чем не посмотреть 

```python
%timeit tags.longest(5)
tags.longest(5)
```

5 самых популярных тэгов:

```python
%timeit tags.most_popular(5)
tags.most_popular(5)
```

```python
%timeit tags.most_words(10)
tags.most_words(10)
```

```python
%timeit tags.most_words_and_longest(10)
tags.most_words_and_longest(10)
```
